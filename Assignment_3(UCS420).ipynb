```python
#Question-6:
import pandas as pd

# Creating sample employee data:
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01",
                     "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}
df = pd.DataFrame(data)
df.to_csv("employees.csv", index=False)
print("employees.csv created successfully!")

df = pd.read_csv("employees.csv")

print("Employees data:")
print(df)

# a)
print(df.shape)

# b)
df.info()

# c)
print(df.describe())

# d)
print(df.head(5))
# for last 3 rows:

print(df.tail(3))
# e) Calculate statistics
# i.
average_salary = df["Salary"].mean()
print("(i) o/p: ", average_salary)

# ii.
total_bonus = df["Bonus"].sum()
print("(ii):", total_bonus)

# iii.
youngest_age = df["Age"].min()
youngest_employee = df[df["Age"] == youngest_age]

print("(iii):")
print(youngest_employee[["Name", "Age"]])

# iv.
highest_rating = df["Rating"].max()
print("(iv) Highest performance rating:", highest_rating)

# f)
df_sorted = df.sort_values(by="Salary", ascending=False)
print(df_sorted)

# g)
# Function :

def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Avg"

df["Performance_Category"] = df["Rating"].apply(performance_category)
print(df[["Name", "Rating", "Performance_Category"]])

# h)
print(df.isnull().sum())

# i)
df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print(df)

# j)
# (i)
more_than_5 = df[df["Years_of_Experience"] > 5]

print("(i) Employees with more than 5 years of exp.:")
print(more_than_5)

# (ii)
it_employees = df[df["Department"] == "IT"]
print("(ii) Employees belonging to IT dept. :")
print(it_employees)

# k)
df["Tax"] = df["Salary"] * 0.10
print(df[["Name", "Salary", "Tax"]])

# l)
df.to_csv("employees_modified.csv", index=False)
```
