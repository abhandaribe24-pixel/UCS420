```python
# Question-1:
roll_no = "1024170427"

L = [int(digit) * 10 for digit in roll_no]

# i.
print("Initial L:", L)

# ii
L.append(50)
print("After append:", L)

L.insert(2, 70)
print("After insert:", L)

# iii.
L.remove(50)
print("After remove:", L)

L.pop(2)
print("After pop:", L)

# iv.
L.sort()
print("Ascending:", L)

L.sort(reverse=True)
print("Descending:", L)

# v.
print("1st three:", L[:3])
print("Last three:", L[-3:])

# vi
average = sum(L) / len(L)

new_list = [x for x in L if x > average]

print("Avg:", average)
print("Elements greater than average:", new_list)
```

```python
# Question-2:
# creating a tuple:
scores = tuple(L[:8])

print("Scores tuple:", scores)

# i.
highest = max(scores)
highest_index = scores.index(highest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)


lowest = min(scores)
lowest_count = scores.count(lowest)

print("Lowest score:", lowest)
print("Lowest score appears:", lowest_count, "times")


# ii.
reversed_scores = list(reversed(scores))

print("Reversed tuple as list:", reversed_scores)


# iii.
user_score = int(input("Enter a score: "))

if user_score in scores:
    print("First occurrence index:", scores.index(user_score))
else:
    print("Score not present in the tuple")


# iv.
# scores[0] = 100  it would not run cz tuples dpont support type assignment

# v.
first_score, second_score, *remaining_scores = scores
```

```python
#Question-3:
import random

#
random.seed(1024170427)

# i.
numbers = [random.randint(100, 900) for _ in range(100)]

print("Generated List:")
print(numbers)

# ii.
odd_numbers = [num for num in numbers if num % 2 != 0]
print("Odd num:")
print(odd_numbers)
print("Count of odd num", len(odd_numbers))

# iii.
even_numbers = [num for num in numbers if num % 2 == 0]
print("\nEven num:")
print(even_numbers)
print("Count of even num:", len(even_numbers))

# iv.
def is_prime(num):
    if num < 2:
        return False

    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False

    return True

prime_number = [num for num in numbers if is_prime(num)]

print("Prime num:")
print(prime_number)
print("Count of prime num:", len(prime_number))

# v.
most_frequent = max(set(numbers), key=numbers.count)
frequency = numbers.count(most_frequent)

print("Most freq. occurring number:", most_frequent)
print("Num of times it occurs:", frequency)
```

```python
#Question-4:
# Same 8 digits of your roll number from Q1
digits = [1, 0, 2, 4, 1, 7, 0, 4]

# Creating sets a+b
A = {digit * 7 for digit in digits}
B = {digit * 9 for digit in digits}

print("Set A:", A)
print("Set B:", B)

# vi.
union = A.union(B)
print("\nUnion of A and B:", union)

# vii.
intersection = A.intersection(B)
print("Intersection of A and B:", intersection)

# viii.
A_minus_B = A.difference(B)
B_minus_A = B.difference(A)

print("A - B:", A_minus_B)
print("B - A:", B_minus_A)

# difference will give elements present in the first set but not the second, while symmetric_diff gives elements present in either set but not both.

# ix.
symmetric_diff = A.symmetric_difference(B)
print("Symmetric difference:", symmetric_diff)

# x.
print("\nIs A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

# xi.
X = int(input("\nEnter a value X to remove from A: "))

A.discard(X)

print("Set A after discard:", A)

# discard is safer option because it does not give an error if X is not present, whereas remove is going to raise a KeyError.
```

```python
# Question-5:
# Original dictionary
my_dict = {
    "name": "Aashna",
    "roll_no": "1024170427",
    "branch": "COPC",
    "age": 19,
    "city": "Patiala"
}

print("Original dictionary:")
print(my_dict)


# i.
my_dict["location"] = my_dict.pop("city")

print("\nAfter renaming city to location:")
print(my_dict)


# ii.
my_dict["cgpa"] = 8.16

print("\nAfter adding CGPA:")
print(my_dict)


# iii.
my_dict["age"] += 1

print("\nAfter increasing age:")
print(my_dict)


# iv.
# Copy 1: using pop()
dict_pop = my_dict.copy()
removed_branch = dict_pop.pop("branch")

print("\nUsing pop():")
print(dict_pop)
print("Removed value:", removed_branch)

# Copy 2: using del
dict_del = my_dict.copy()
del dict_del["branch"]

print("\nUsing del:")
print(dict_del)

# pop removes the key & returns its value, while del only removes the key and does not return its value.
# v.
print("Key-value pairs are:")

for key, value in my_dict.items():
    print(f"{key} → {value}")


# vi.
if "email" in my_dict:
    print("\nEmail:", my_dict["email"])
else:
    print("\nEmail key does not exist.")


# vii.
friend_dict = {
    "name": "XYZ",
    "roll_no": "1234567",
    "branch": "COPC",
    "age": 20,
    "city": "Delhi"
}

# Merge using **dict1, **dict2
merged_dict = {**my_dict, **friend_dict}

print("Friend dictionary:")
print(friend_dict)

print("Merged dictionary:")
print(merged_dict)

# When both dictionaries have the same key, the value from friend_dict wins because it comes second.


# viii.
string_values = {
    key: value
    for key, value in my_dict.items()
    if isinstance(value, str)
}
print("Dictionary containing only string values:")
print(string_values)
```
