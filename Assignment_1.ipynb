```python
def add_odd(n):
    total = 0

    for i in range(1, n + 1):
        if i % 2 != 0:
            total = total + i

    return total


n = int(input("Enter n: "))
print("Sum of odd numbers =", add_odd(n))
```

```python
def sum_prime(n):
    total = 0

    for i in range(2, n + 1):
        prime = True

        for j in range(2, i):
            if i % j == 0:
                prime = False
                break

        if prime:
            total = total + i

    return total


n = int(input("Enter n: "))
print("Sum of prime numbers =", sum_prime(n))
```
